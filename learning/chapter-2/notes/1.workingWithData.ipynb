# Converting tokens into token IDs

- before encoding need to convert the tokens into vocabulary

![2. Text Vocab](../images/2.tokens-to-vocab.png)

```python
all_words = sorted(list(set(result)))
all_words
```

```python
vocab = {token: integer for integer, token in enumerate(all_words) }
vocab
```

```python
# making a simple tokenizer
class simpleTokenizerV1:
    def __init__(self, vocab):
        self.stoi = vocab
        self.itos = {token: s for s, token in vocab.items()}

    def encode(self, text):
        # convert into tokens and then assign the id's from vocab
        preprocessed = re.split(r'([,.?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        return [self.stoi[pre] for pre in preprocessed]

    def decode(self, ids):
        text = " ".join([self.itos[i] for i in ids])
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text) 
        return text
```

![2. Text Vocab](../images/3.tokenizers_simple.png)

```python
tokenizer = simpleTokenizerV1(vocab)
text = """"It's the last he painted, you know,"
 Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
tokenizer.decode(ids)
```

```python
# new tokenizer with handling unknown tokens
class simpleTokenizerV2:
    def __init__(self, vocab):
        self.stoi = vocab
        self.itos = {token: s for s, token in vocab.items()}

    def encode(self, text):
        # convert into tokens and then assign the id's from vocab handle the missing tokens too
        preprocessed = re.split(r'([,.?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        preprocessed = [item if item in self.stoi else "<|unk|>" for item in preprocessed] # loop , condition and adding into list at a time
        return [self.stoi[pre] for pre in preprocessed]

    def decode(self, ids):
        text = " ".join([self.itos[i] for i in ids])
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text) 
        return text
```

```python
tokenizer = simpleTokenizerV2(vocab)
text = """"It's harsha the last he painted, you know,"
 Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
tokenizer.decode(ids)
```

```python
import torch

from torch.utils.data import Dataset, DataLoader

class GPTDatasetV1(Dataset):
    def __init__(self, text, tokenizer, stride, max_length):
        self.input_ids=[]
        self.target_ids=[]

        token_ids = tokenizer.encode(text)
        # from these token ids create input ids and thier targets.

        for i in range(0, len(token_ids)-max_length, stride):
            input_chunk  = token_ids[i: i+max_length] # at a time taking maxlength of inputs and thier target not a single
            target_chunk = token_ids[i+1: i+max_length+1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __len__(self):
        return len(self.input_ids) # returns all rows in the dataset

    def __getitem__(self, idx):
        return self.input_ids[idx], self.target_ids[idx] #Returns a single row from the dataset
```

```python
# dataloader needs above dataset

def create_dataloader_v1(txt, max_length=256, stride=128, batch_size=4,shuffle=True, drop_last=True, num_workers=0):
    tokenizer = tiktoken.get_encoding("gpt2")
    dataset = GPTDatasetV1(txt, tokenizer, stride, max_length)
    dataloader = DataLoader(
    dataset,
    batch_size=batch_size,
    shuffle=shuffle,
    drop_last=drop_last,     #drop_last=True drops the last batch if it is shorter than the specified batch_size to prevent loss spikes during training
    num_workers=num_workers # no of cpu processes to use for preprocessing
    )

    return dataloader
    
    
```

- `Dataset` = "how to get one sample"
- `DataLoader` = "how to get many samples efficiently (batching, shuffling, parallelism)"
- `Stride` = Each window shifts by 1 token (stride=1), so consecutive windows overlap heavily.
 -  max_length =4, stride =1
- Text:  [a, b, c, d, e, f, g, h, i, j]

- Window 0:  [a, b, c, d]     → target: [b, c, d, e]
- Window 1:  [b, c, d, e]     → target: [c, d, e, f]
- Window 2:  [c, d, e, f]     → target: [d, e, f, g]
- Window 3:  [d, e, f, g]     → target: [e, f, g, h]

```python
print(len(raw_text))
dataloader1 = create_dataloader_v1(raw_text, max_length=4, stride=1, batch_size=1, shuffle=False)
data_iter = iter(dataloader1)
first_batch = next(data_iter)
print(first_batch)
```

# Encoding word positions

- However, a minor
shortcoming of LLMs is that their self-attention mechanism (see chapter 3) doesn’t
have a notion of position or order for the tokens within a sequence. The way the previously introduced embedding layer works is that the same token ID always gets
mapped to the same vector representation, regardless of where the token ID is positioned in the input sequence.

  ![8. Embedding positions](../images/8.embeddingpositions.png)


- The embedding layer converts a token ID into the same vector
representation regardless of where it is located in the input sequence

  ![9. Embedding positions](../images/9.embedding_2.png)

- it is helpful to inject additional position information into the LLM.

- To achieve this, we can use two broad categories of position-aware embeddings: relative positional embeddings and absolute positional embeddings. Absolute positional embeddings are directly associated with specific positions in a sequence.

- For each position in the input sequence, a unique embedding is added to the token’s embedding to
convey its exact location.

# absolute positional embeddings

- Positional embeddings are vectors added to token embeddings in Transformer models to encode the sequence order of input data, compensating for the self-attention mechanism's inherent permutation invariance.  Without these embeddings, Transformers would process inputs as unordered sets, losing critical contextual relationships like word order in text or spatial arrangement in images.

-   ![10. Embedding positions](../images/10.positionalEmb.png)

# relative positional embeddings

- the emphasis of relative positional embeddings is on the relative position or distance between tokens. This means
the model learns the relationships in terms of “how far apart” rather than “at which
exact position.” The advantage here is that the model can generalize better to sequences
of varying lengths, even if it hasn’t seen such lengths during training.
