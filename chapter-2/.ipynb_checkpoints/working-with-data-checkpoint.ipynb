```python
import os

with open("verdict.txt","r", encoding="utf-8") as f:
    data_raw_text = f.read()
```

## converting text into tokens
- need clean it remove the \n, extra spaces, in normal tokenizers these are not the tokens


```python
import re

result = re.split(r'([,.:;?_!"()\']|--|\s)', data_raw_text) #just splitting
result = [item for item in result if item.strip()]
preprocessed = result
preprocessed
print(len(preprocessed))
```

## converting tokens into token id's
- we are building a vocabulary by converting entire text into tokens
- These are alphabetically sorted and no duplicates with each have to assign with unique id's.

```python
vocabulary = sorted(set(preprocessed))
len(vocabulary)
```

```python
tokens = {token: integer for integer, token in enumerate(vocabulary) }
len(tokens)
```

```python
class simpleTokenizerV1:
    def __init__(self, tokens):
        self.stoi = tokens
        self.itos = {i:s for s, i in tokens.items()}

    def encode(self, text):
        # covert it into tokens and assign existing id's
        # print(f"received text is : {text}")
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item for item in preprocessed if item.strip()]
        
        ids= [self.stoi[s] for s in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.itos[i] for i in ids])
        # replace space before the some special symbols
        preprocessed = re.sub(r'\s+([,.?!"()\'])', r'\1' ,text)
        return text
        
        
```

```python
tokenizer = simpleTokenizerV1(tokens)
text = """prodigious phrase about the honour being _mine_--oh, I
was princely, my dear Rickham! I was posing to myself like
one of my own sitters"""
ids = tokenizer.encode(text)
print(ids)
```

```python
tokenizer.decode(ids)
```

## Adding special context tokens
- we are adding special tokens to handle the unknown words in the context or in given question , <|unk|> and <|endoftext|> token that we can use to separate two unrelated text sources.

```python
vocabulary.extend(["<|endoftext|>","<|unk|>"])
```

```python
tokens = {token: integer for integer, token in enumerate(vocabulary) }
len(tokens)
```

```python
class simpleTokenizerV2:
    def __init__(self, tokens):
        self.stoi = tokens
        self.itos = {i:s for s, i in tokens.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item for item in preprocessed if item.strip()]

        preprocessed = [ item if item in self.stoi
                       else "<|unk|>" for item in preprocessed
                       ]
        
        ids= [self.stoi[s] for s in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.itos[i] for i in ids])
        preprocessed = re.sub(r'\s+([,.?!"()\'])', r'\1' ,text)
        return text
        
        
```

```python
tokenizer = simpleTokenizerV2(tokens)
text = """Hi harsha bhaai , how are you?"""
ids = tokenizer.encode(text)
tokenizer.decode(ids)
```

## Data sampling with a sliding window
- most important how can we predict next token based on the given input

```python
enc_text = tokenizer.encode(data_raw_text)
len(enc_text)
```

```python
# defining the targets for the each inputs.
context_size = 4
x = enc_text[:context_size]
y = enc_text[1:context_size+1]

print (f"the x context is {x}")
print (f"the y context is {y}")
```

```python
for i in range(1, context_size+1):
    context = enc_text[:i]
    decode = enc_text[i]

    print(f"the input is {tokenizer.decode(context)} -----> decode {tokenizer.decode([decode])}")
```
